# Geometry-controlled magnon-polariton excitations in a bilayer planar cavity

The cavity transmission amplitude for the bilayer problem is extracted using the transfer matrix method. Let film 1 have thickness $d_1$, internal wave number $k_1$, mismatch parameter $\beta_1$, and isolated-slab amplitudes $r_1, t_1$. Let film 2 have thickness $d_2$, internal wave number $k_2$, mismatch parameter $\beta_2$, and isolated-slab amplitudes $r_2, t_2$. The two films are separated by a nonmagnetic spacer of length $s$, and both are embedded in the same cavity medium. The isolated-slab amplitudes for each film are therefore:

$$r_{\nu} = \frac{\beta_{\nu}\left(1-e^{2ik_{\nu}d_{\nu}}\right)}{1-\beta_{\nu}^2 e^{2ik_{\nu}d_{\nu}}}, \qquad t_{\nu} = \frac{(1-\beta_{\nu}^2)e^{ik_{\nu}d_{\nu}}}{1-\beta_{\nu}^2 e^{2ik_{\nu}d_{\nu}}}, \qquad \nu=1,2.$$

When two such scatterers are placed in series inside the same background medium, their multiple reflections across the spacer $s$ can be summed exactly. The effective bilayer transmission amplitude is:

$$t_{\mathrm{eff}} = \frac{t_1 t_2 e^{iqs}}{1-r_1 r_2 e^{2iqs}}.$$

If the two films are not identical, the bilayer composite is generally not left-right symmetric, and one must distinguish the effective reflection amplitudes for waves incident from the left and from the right:

$$r_{\mathrm{eff}}^{L} = r_1+\frac{t_1^{\,2} r_2 e^{2iqs}}{1-r_1 r_2 e^{2iqs}}, \qquad r_{\mathrm{eff}}^{R} = r_2+\frac{t_2^{\,2} r_1 e^{2iqs}}{1-r_1 r_2 e^{2iqs}}.$$

These formulas are the exact two-scatterer composition laws for reciprocal slabs embedded in the same medium. If the two films are identical, then $r_{\mathrm{eff}}^{L}=r_{\mathrm{eff}}^{R}$, and the effective bilayer becomes left-right symmetric.

To insert the effective bilayer back into the cavity, it is convenient to use a transfer-matrix representation [38, 39]. For a reciprocal but generally asymmetric scatterer with left reflection amplitude $r^L$, right reflection amplitude $r^R$, and transmission amplitude $t$, the transfer matrix may be written as:

$$M_{\mathrm{asym}}(r^L,r^R,t) = \frac{1}{t} \begin{pmatrix} t^2-r^L r^R & r^R \\ -r^L & 1 \end{pmatrix}.$$

When the scatterer is left-right symmetric, this reduces to the familiar form:

$$M_{\mathrm{sym}}(r,t) = \frac{1}{t} \begin{pmatrix} t^2-r^2 & r \\ -r & 1 \end{pmatrix}.$$

The free propagation matrix through a cavity spacer of length $\ell$ is:

$$P(q,\ell) = \begin{pmatrix} e^{iq\ell} & 0 \\ 0 & e^{-iq\ell} \end{pmatrix}.$$

The cavity walls are symmetric scatterers, so each wall is represented by $M_{\mathrm{wall}} = M_{\mathrm{sym}}(r_c,t_c)$, while the effective bilayer is represented by $M_{\mathrm{eff}} = M_{\mathrm{asym}}\!\left(r_{\mathrm{eff}}^{L},r_{\mathrm{eff}}^{R},t_{\mathrm{eff}}\right)$.

Using the symmetric cavity geometry, the total transfer matrix of the full bilayer cavity is:

$$M_{\mathrm{tot}} = M_{\mathrm{wall}} \,P(q,\ell)\, M_{\mathrm{eff}} \,P(q,\ell)\, M_{\mathrm{wall}}$$

where $\ell=\frac{L-d_1-d_2-s}{2}$. The cavity transmission amplitude is then extracted from the transfer matrix as:

$$S_{\mathrm{bi}}(\omega,H) = \frac{\det M_{\mathrm{tot}}}{\left(M_{\mathrm{tot}}\right)_{22}}.$$

This equation is used for both the symmetric and asymmetric bilayer cases in the macrospin limit ($J=0$). For the asymmetric case, we apply a differential bias field:

$$H_1 = H_0 + \frac{\delta H}{2}, \qquad H_2 = H_0 - \frac{\delta H}{2}$$

For the exchange-driven regime ($J\neq 0$), the transmission is formulated using the standard Tavis-Cummings model:

$$S_{n,\mathrm{bi}}(\omega) = \frac{\kappa_{c,n}}{i(\omega-\omega_{c,n})-\kappa_{c,n} - i\displaystyle\sum_{p\in\mathcal O}\Sigma_p(\omega)}$$

where the self-energy for each odd standing-spin-wave family $p$ is:

$$\Sigma_p(\omega) = \mathbf g_p^{\,T} \left[ \omega I - \Omega_p + iK_p \right]^{-1} \mathbf g_p$$

with the damping matrix $K_p$, the coupling vector $\mathbf g_p$, and the effective $2 \times 2$ magnetic mode matrix $\Omega_p$ defined as:

$$K_p = \begin{pmatrix} \kappa_{n,1}^{(p)} & 0 \\ 0 & \kappa_{n,2}^{(p)} \end{pmatrix}, \qquad \mathbf g_p = \begin{pmatrix} g_{n,1}^{(p)} \\ g_{n,2}^{(p)} \end{pmatrix}, \qquad \Omega_p = \begin{pmatrix} \omega_{\mathrm{SWR},1}^{(p)} & J^{(p)}_{\mathrm{int}} \\ J^{(p)}_{\mathrm{int}} & \omega_{\mathrm{SWR},2}^{(p)} \end{pmatrix}$$

and the standing-spin-wave resonance frequencies given by:

$$\omega_{\mathrm{SWR},\nu}^{(p)} = \sqrt{\left[ \omega_{H,\nu} + 2J\omega_{M,\nu}\left(\frac{p\pi}{d_\nu}\right)^2 \right] \left[ \omega_{M,\nu} + \omega_{H,\nu} + 2J\omega_{M,\nu}\left(\frac{p\pi}{d_\nu}\right)^2 \right]}$$

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from matplotlib.ticker import MultipleLocator
from pathlib import Path
import scienceplots

plt.style.use(['science', 'no-latex'])
plt.rcParams.update({'font.size': 16, 'axes.labelsize': 16, 'axes.titlesize': 16, 
                     'xtick.labelsize': 16, 'ytick.labelsize': 16, 'legend.fontsize': 16})

outdir = Path(".")

# ============================================================
# Physical Parameters
# ============================================================
c0 = 299792458.0
gamma_2pi = 28e9
gamma = 2 * np.pi * gamma_2pi
mu0_Ms = 0.175
omega_M = gamma * mu0_Ms
eta = 15.0
L = 46e-3
alpha = 3e-4
J_ex = 3e-16 
fc = 9.84e9
wc = 2 * np.pi * fc
wc_thick_3 = 2 * np.pi * 10.03e9
wc_thick_4 = 2 * np.pi * 11.27e9
kappa_c = 2 * np.pi * 1.44e6
g1_ref = 2 * np.pi * 28e6 

# ============================================================
# Function
# ============================================================
def csqrt(z):
    return np.lib.scimath.sqrt(z)

def q_of_w(w):
    return w / c0

def omega_H(H_field):
    return gamma * H_field

def omega_fmr(H_field):
    wH = omega_H(H_field)
    return np.sqrt(wH * (wH + omega_M))

def k_macrospin(w, H_field, eta_=eta, alpha_=alpha, omega_M_=omega_M, gamma_=gamma):
    q = q_of_w(w)
    k_eps = np.sqrt(eta_) * q
    wH = gamma_ * H_field
    omega_k = wH - 1j * alpha_ * w
    denom = omega_k**2 - w**2
    u = omega_k * omega_M_ / denom
    v = w * omega_M_ / denom
    return k_eps * csqrt(1 + u - v**2 / (1 + u))

def wall_coeffs(w, Delta):
    q = q_of_w(w)
    tc = 1j / (1j + q * Delta)
    rc = -q * Delta / (1j + q * Delta)
    return tc, rc

def beta_coeff(w, H_field, eta_=eta, alpha_=alpha, omega_M_=omega_M, gamma_=gamma):
    q = q_of_w(w)
    k = k_macrospin(w, H_field, eta_=eta_, alpha_=alpha_, omega_M_=omega_M_, gamma_=gamma_)
    return (eta_ * q - k) / (eta_ * q + k)

def slab_rt(w, H_field, d, eta_=eta, alpha_=alpha, omega_M_=omega_M, gamma_=gamma):
    k = k_macrospin(w, H_field, eta_=eta_, alpha_=alpha_, omega_M_=omega_M_, gamma_=gamma_)
    beta = beta_coeff(w, H_field, eta_=eta_, alpha_=alpha_, omega_M_=omega_M_, gamma_=gamma_)
    denom = 1.0 - beta**2 * np.exp(2j * k * d)
    r = beta * (1.0 - np.exp(2j * k * d)) / denom
    t = (1.0 - beta**2) * np.exp(1j * k * d) / denom
    return r, t

def compose_two_scatterers_same_medium(r1, t1, r2, t2, q, s):
    phase1 = np.exp(1j * q * s)
    phase2 = np.exp(2j * q * s)
    denom = 1.0 - r1 * r2 * phase2
    r_eff_L = r1 + (t1**2) * r2 * phase2 / denom
    r_eff_R = r2 + (t2**2) * r1 * phase2 / denom
    t_eff = t1 * t2 * phase1 / denom
    return r_eff_L, r_eff_R, t_eff

def M_scatter(rL, rR, t):
    return np.array([[(t**2 - rL * rR) / t, rR / t], [-rL / t, 1.0 / t]], dtype=complex)

def M_prop(q, ell):
    return np.array([[np.exp(1j * q * ell), 0.0], [0.0, np.exp(-1j * q * ell)]], dtype=complex)

def transmission_from_M(M):
    detM = np.linalg.det(M)
    return detM / M[1, 1]

def S_cao_eq9(w, H_field, d, Delta):
    q = q_of_w(w)
    k = k_macrospin(w, H_field)
    beta = beta_coeff(w, H_field)
    tc, rc = wall_coeffs(w, Delta)
    phi = q * (L - d)
    num = (1.0 - beta**2) * tc**2 * np.exp(1j * (k - q) * d)
    den = (1.0 - beta * rc * np.exp(1j * phi)) ** 2 - np.exp(2j * k * d) * (beta - rc * np.exp(1j * phi)) ** 2
    return num / den

def S_bilayer_transfer(w, H_field, d1, d2, s, Delta):
    ell = 0.5 * (L - d1 - d2 - s)
    if np.real(ell) < -1e-18:
        raise ValueError("Invalid geometry: d1 + d2 + s > L.")
    q = q_of_w(w)
    tc, rc = wall_coeffs(w, Delta)
    Mw = M_scatter(rc, rc, tc)

    r1, t1 = slab_rt(w, H_field, d1)
    r2, t2 = slab_rt(w, H_field, d2)
    r_eff_L, r_eff_R, t_eff = compose_two_scatterers_same_medium(r1, t1, r2, t2, q, s)
    Meff = M_scatter(r_eff_L, r_eff_R, t_eff)

    Mtot = Mw @ M_prop(q, ell) @ Meff @ M_prop(q, ell) @ Mw
    return transmission_from_M(Mtot)

def S_bilayer_asym(w, H0, d1, d2, s, Delta, dH=0.0):
    ell = 0.5 * (L - d1 - d2 - s)
    if np.real(ell) < -1e-18:
        return np.nan + 1j * np.nan
    H1 = H0 + dH / 2.0
    H2 = H0 - dH / 2.0
    q = q_of_w(w)
    tc, rc = wall_coeffs(w, Delta)
    r1, t1 = slab_rt(w, H1, d1)
    r2, t2 = slab_rt(w, H2, d2)
    rL, rR, t_eff = compose_two_scatterers_same_medium(r1, t1, r2, t2, q, s)

    Mw = M_scatter(rc, rc, tc)
    Meff = M_scatter(rL, rR, t_eff)
    P = M_prop(q, ell)
    Mtot = Mw @ P @ Meff @ P @ Mw
    return transmission_from_M(Mtot)

def map_from_model(fvals, Hvals, model):
    out = np.zeros((len(Hvals), len(fvals)), dtype=complex)
    wvals = 2 * np.pi * fvals
    for i, H0 in enumerate(Hvals):
        for j, ww in enumerate(wvals):
            out[i, j] = model(ww, H0)
    P = np.abs(out) ** 2
    P = np.maximum(P, np.max(P) * 1e-14)
    return 10.0 * np.log10(P / np.max(P))

def linecut_from_model(fvals, model):
    wvals = 2 * np.pi * fvals
    out = np.zeros(len(fvals), dtype=complex)
    for j, ww in enumerate(wvals):
        out[j] = model(ww)
    P = np.abs(out) ** 2
    P = np.maximum(P, np.max(P) * 1e-14)
    db = 10.0 * np.log10(P / np.max(P))
    return db, P

def extract_g(S_line, fvals):
    P = np.abs(S_line)**2
    dP = np.diff(P)
    idx = np.where((dP[:-1] > 0) & (dP[1:] < 0))[0] + 1
    if len(idx) < 2:
        return np.nan
    top = idx[np.argsort(P[idx])[-2:]]
    top = np.sort(top)
    split = fvals[top[-1]] - fvals[top[0]]
    return 0.5 * split

def power_db(S):
    P = np.abs(S) ** 2
    P = np.maximum(P, np.max(P) * 1e-14)
    return 10.0 * np.log10(P / np.max(P))

def omega_swr(H_field, p, d):
    omega_H_local = gamma * H_field
    exch = 2.0 * J_ex * omega_M * (p * np.pi / d) ** 2
    return np.sqrt((omega_H_local + exch) * (omega_M + omega_H_local + exch))

def kappa_swr(omega_p):
    return alpha * omega_p

def H_res_p(p, d):
    exch = 2.0 * J_ex * omega_M * (p * np.pi / d) ** 2
    return (-omega_M + np.sqrt(omega_M**2 + 4.0 * wc**2)) / (2.0 * gamma) - (exch / gamma)

def single_film_self_energy(W, H_field, d, p_list):
    Hcol = H_field[:, None]
    W2 = W[None, :]
    Sigma = np.zeros((len(H_field), len(W)), dtype=complex)
    for p in p_list:
        wp = omega_swr(Hcol, p, d)
        kp = kappa_swr(wp)
        gp = g1_ref * np.sqrt(d / 1e-6) / p
        Sigma += gp**2 / (W2 - wp + 1j * kp)
    return Sigma

def bilayer_self_energy(W, H_field, d1, d2, p_list, dH=0.0, eps_g=0.0, Jint0_MHz=0.0):
    H1 = H_field[:, None] + dH / 2.0
    H2 = H_field[:, None] - dH / 2.0
    W2 = W[None, :]
    Sigma = np.zeros((len(H_field), len(W)), dtype=complex)

    for p in p_list:
        w1 = omega_swr(H1, p, d1)
        w2 = omega_swr(H2, p, d2)
        k1 = kappa_swr(w1)
        k2 = kappa_swr(w2)

        g1 = g1_ref * np.sqrt(d1 / 1e-6) * (1.0 + eps_g) / p
        g2 = g1_ref * np.sqrt(d2 / 1e-6) * (1.0 - eps_g) / p
        Jp = 2 * np.pi * Jint0_MHz * 1e6 / p

        a11 = W2 - w1 + 1j * k1
        a22 = W2 - w2 + 1j * k2
        det = a11 * a22 - Jp**2

        Sigma += (g1**2 * a22 + g2**2 * a11 + 2.0 * g1 * g2 * Jp) / det
    return Sigma

def transmission_from_sigma(W, Sigma):
    W2 = W[None, :]
    return kappa_c / (1j * (W2 - wc) - kappa_c - 1j * Sigma)

def family_mode_matrix(H0, p, d, dH=0.0, eps_g=0.0, Jint0_MHz=0.0):
    H1 = H0 + dH / 2.0
    H2 = H0 - dH / 2.0
    w1 = omega_swr(H1, p, d)
    w2 = omega_swr(H2, p, d)
    k1 = kappa_swr(w1)
    k2 = kappa_swr(w2)
    g1 = g1_ref * np.sqrt(d / 1e-6) * (1.0 + eps_g) / p
    g2 = g1_ref * np.sqrt(d / 1e-6) * (1.0 - eps_g) / p
    Jp = 2 * np.pi * Jint0_MHz * 1e6 / p

    M = np.array(
        [[wc - 1j * kappa_c, g1, g2], [g1, w1 - 1j * k1, Jp], [g2, Jp, w2 - 1j * k2]],
        dtype=complex,
    )
    return M

def mode_metrics(H0, p, d, dH=0.0, eps_g=0.0, Jint0_MHz=0.0):
    M = family_mode_matrix(H0, p, d, dH=dH, eps_g=eps_g, Jint0_MHz=Jint0_MHz)
    vals, vecs = np.linalg.eig(M)
    idx = np.argsort(vals.real)
    vals = vals[idx]
    vecs = vecs[:, idx]

    cavw = []
    for j in range(3):
        v = vecs[:, j]
        norm = np.sum(np.abs(v) ** 2)
        cavw.append(float(np.abs(v[0]) ** 2 / norm))
    cavw = np.array(cavw)

    middle_weight = cavw[1]
    outer_split = vals[2].real - vals[0].real
    min_sep = min(vals[1].real - vals[0].real, vals[2].real - vals[1].real)
    return middle_weight, outer_split, min_sep, vals, cavw

def family_self_energy(W, H_field, p, d1, d2, dH=0.0, eps_g=0.0, Jint0_MHz=0.0):
    H1 = H_field[:, None] + dH / 2.0
    H2 = H_field[:, None] - dH / 2.0
    W2 = W[None, :]

    w1 = omega_swr(H1, p, d1)
    w2 = omega_swr(H2, p, d2)
    k1 = kappa_swr(w1)
    k2 = kappa_swr(w2)

    g1 = g1_ref * np.sqrt(d1 / 1e-6) * (1.0 + eps_g) / p
    g2 = g1_ref * np.sqrt(d2 / 1e-6) * (1.0 - eps_g) / p
    Jp = 2 * np.pi * Jint0_MHz * 1e6 / p

    a11 = W2 - w1 + 1j * k1
    a22 = W2 - w2 + 1j * k2
    det = a11 * a22 - Jp**2

    Sigma = (g1**2 * a22 + g2**2 * a11 + 2.0 * g1 * g2 * Jp) / det
    return Sigma
```
